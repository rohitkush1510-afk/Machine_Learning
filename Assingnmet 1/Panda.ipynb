## 15. Displaying a DataFrame (`df`, `head()`, `tail()`)

**Description:** Writing `df` alone shows the whole frame (with `...` in the middle). `head()` peeks at the top rows, `tail()` at the bottom — quick sanity checks after loading.

```python
# Impl 1: full display + default head/tail (first/last 5 rows)
print(df)
print()
print(df.head())
print()
print(df.tail())               # your cell [10]
```

```python
# Impl 2: custom number of rows
print(df.head(10))             # first 10
print()
print(df.tail(3))              # last 3
```

```python
# Impl 3: programmatic access to the rows
first_rows = df.iloc[:5]      # slice-based head
last_row   = df.iloc[-1:]     # just the final record
print(first_rows)
print()
print(last_row)
```

## 16. `df.info()` — Metadata summary

**Description:** Prints the class, index range, every column's non-null count and dtype, plus memory usage — your screenshot showing 150 entries, 6 columns, dtypes `float64`/`int64`/`str`. Best first command after loading data (finds missing values instantly).

```python
# Impl 1: standard call
df.info()
```

```python
# Impl 2: limit displayed rows (useful for huge datasets)
df.info(max_cols=3)
```

```python
# Impl 3: rebuild the same info manually
for c in df.columns:
    print(c, "| non-null:", df[c].notna().sum(), "| dtype:", df[c].dtype)
```

## 17. `df.describe()` — Statistical summary

**Description:** Auto-computes count, mean, std, min, 25%, 50% (median), 75%, max for all numeric columns — exactly the table in your cell [13].

```python
# Impl 1: default (numeric columns only)
df.describe()
```

```python
# Impl 2: include text columns too (gives count/unique/top/freq)
df.describe(include="all")
```

```python
# Impl 3: pick specific statistics
print(df.describe(percentiles=[.10, .90]))           # add 10th & 90th percentile
print()
print(df[["SepalLengthCm","PetalWidthCm"]].describe())   # subset of columns
```

## 20. `df.drop()` — Removing columns or rows

**Description:** Deletes specified columns (`columns=`) or rows (`index=`). By default it returns a new DataFrame (original untouched) unless you pass `inplace=True`. Your cell [20] dropped the `Id` column.

```python

df.drop(columns="Id").head()
print("df still has 'Id':", "Id" in df.columns)
```

## 21. Column selection & filtering (`df[col]`, `df.loc`, `df.iloc`)

**Description:** Grabbing one column gives a `Series`; boolean filters like `df[df.Class=='A']` give matching rows. `loc` selects by label, `iloc` by integer position.

```python

print(df2["name"])                                  
print()
print(df2[df2["class"] == "A"])                     
```

```python

df.loc[0:5, ["SepalLengthCm","Species"]]    
```

```python

df.iloc[0:5, 0:3]
```
